# LSTM Forecaster Benchmarking & Baseline Comparisons

This notebook presents an independent comparative evaluation of the **Stacked LSTM Point Forecaster**
against standard empirical baselines:
1. **Persistence Baseline** (24-hour diurnal repetition)
2. **Ridge Regression** (L2-regularized linear model on flattened sequence windows)
3. **Stacked LSTM Forecaster** (Sequence-to-sequence deep neural architecture)

Evaluations enforce a strict **chronological 70% Train / 10% Validation / 20% Test split** to eliminate test leakage.

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

if os.path.basename(os.getcwd()) == 'notebooks':
    os.chdir('..')

print('Working Directory:', os.getcwd())
```

## 1. Run Baseline Evaluation & Empirical Interval Calibration
Executes `backend.ai.lstm.evaluate.run_evaluation()` to benchmark on held-out test windows.

```python
from backend.ai.lstm.evaluate import run_evaluation

report = run_evaluation()

print('=== SOLAR FORECAST METRICS ===')
print('Persistence:     ', report['solar_evaluation']['persistence'])
print('Ridge Regression:', report['solar_evaluation']['ridge_regression'])
print('Stacked LSTM:    ', report['solar_evaluation']['lstm'])
print('Skill vs Persist:', report['solar_evaluation']['lstm_skill_vs_persistence'])
print('Interval Coverage (PICP):', report['solar_evaluation']['prediction_intervals']['picp_observed_pct'], '%')

print('\n=== BUILDING LOAD FORECAST METRICS ===')
print('Persistence:     ', report['load_evaluation']['persistence'])
print('Ridge Regression:', report['load_evaluation']['ridge_regression'])
print('Stacked LSTM:    ', report['load_evaluation']['lstm'])
print('Skill vs Persist:', report['load_evaluation']['lstm_skill_vs_persistence'])
print('Interval Coverage (PICP):', report['load_evaluation']['prediction_intervals']['picp_observed_pct'], '%')
```
